```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Create sample data using numpy
np.random.seed(42)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = np.random.randint(50, 150, size=6)
expenses = np.random.randint(30, 100, size=6)

# Create a pandas DataFrame
df = pd.DataFrame({
    'Month': months,
    'Sales': sales,
    'Expenses': expenses
})

# Calculate profit
df['Profit'] = df['Sales'] - df['Expenses']

# Display the DataFrame
print(df)
print(f"\nTotal Profit: ${df['Profit'].sum()}")

# Create a visualization with matplotlib
fig, ax = plt.subplots(figsize=(10, 6))

x = np.arange(len(months))
width = 0.35

bars1 = ax.bar(x - width/2, df['Sales'], width, label='Sales', color='skyblue')
bars2 = ax.bar(x + width/2, df['Expenses'], width, label='Expenses', color='salmon')

ax.set_xlabel('Month')
ax.set_ylabel('Amount ($)')
ax.set_title('Monthly Sales vs Expenses')
ax.set_xticks(x)
ax.set_xticklabels(months)
ax.legend()
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

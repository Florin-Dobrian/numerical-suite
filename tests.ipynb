```python
import numpy as np
```

```python
A = np.array([[2, 1], [1, 3]])
b = np.array([5, 7])
```

```python
x = np.linalg.solve(A, b)
```

```python
print("Solution to Ax = b:", x)
print("Verify A @ x:", A @ x)
```
